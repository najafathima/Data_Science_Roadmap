#### <b> CONVOLUTIONAL NEURAL NETWORK (CNN)</b>

A Convolutional Neural Network (CNN) is a type of neural network designed specifically to process grid-like data such as images.

Instead of connecting every neuron to every pixel (like a normal neural network), CNNs use filters (kernels) that slide over the image to detect patterns like:

- edges
- textures
- shapes
- objects

example tasks:
- image classification(cat vs dog)
- face recognition
- medical image diagnosis
- self-driving car vision

A typical CNN look like :

Image

  ↓

Convolution Layer

  ↓

Activation (ReLU)

  ↓

Pooling Layer

  ↓

Convolution

  ↓

Pooling

  ↓

Flatten

  ↓

Fully Connected Layer

  ↓

Output

#### Convolution Opertaion:

Example image matrix: 5 x 5

kernel : 3 x 3

the keranl slides across the image and performs:

element-wise multiplication + sum

result = Feature map

in the given image , the kernel is for detecting/understanding horizontal edge, likewise there will be filters/kernels for verticals, rounded etc edges and other feature extraction

to get the size of the resultant matrix:

n-f+1 

n= size of the image

f= size of the kernel

eg: 6 - 3 + 1 = 4 (as in the above image)

But , when we give a 6x6 input , we are getting a 4x4 output, that means image size decreasing -> loosing some information

To prevent this loose of innformation, we use **PADDING**

#### *PADDING:*

When a convolution filter (kernel) moves across an image, it cannot fully cover the pixels at the edges. Because of this, the output feature map becomes smaller than the input image.

eg: 
- input image size= 6x6
- kernel size= 3x3
- without padding , the o/p is: 3x3

Padding adds extra pixels around the image:

eg:
- original image : 6x6
-  after adding 1 layer of padding on each side: 8x8
 -> 
$$
\begin{bmatrix}
0 & 0 & 0 & 0 & 0 & 0 & 0 & 0 \\
0 & x & x & x & x & x & x & 0 \\
0 & x & x & x & x & x & x & 0 \\
0 & x & x & x & x & x & x & 0 \\
0 & x & x & x & x & x & x & 0 \\
0 & x & x & x & x & x & x & 0 \\
0 & x & x & x & x & x & x & 0 \\
0 & 0 & 0 & 0 & 0 & 0 & 0 & 0
\end{bmatrix}
$$

- the o/p will be = 6x6

thats : n+2p-f+1 = 6+2-3+1 = 6
