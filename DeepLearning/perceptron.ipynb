### **DEEP LEARNING**

Deep Learning is a subset of machine learning that uses multi-layered artificial neural networks, inspired by the human brain, to automatically learn hierarchical representations from large amounts of data. It excels in processing unstructured data like images and text by transforming inputs through successive layers into complex features without manual engineering.

### <u><B> *PERCEPTRON*</B></U>

A perceptron is the simplest form of an artificial neural network, acting as a binary linear classifier that mimics a biological neuron.

It is just:

Weighted sum + Bias + Activation

<img src="Perceptron-model-depicting-its-key-components-inputs-weights-bias-and-output.png" width=400/>

A perceptron is a simple binary classifier algorithm using a single artificial neuron with a step activation function to separate linearly separable data via weighted inputs and a bias

Single-layer perceptrons cannot solve non-linearly separable problems like XOR, as they are restricted to linear decision boundaries.

**MLP for Non-Linear Problems8**

Multilayer Perceptrons (MLPs) solve nonlinear problems by stacking hidden layers with non-linear activations, enabling representation of complex decision boundaries through hierarchical feature learning.

### Perceptron Model

A perceptron takes multiple input features x_1, x_2, ... , x_n ,  
each multiplied by a learnable weight w_1, w_2,..., w_n,  
and added with a bias term b 

The linear combination is computed as:
\[
z = \sum_{i=1}^{n} w_i x_i + b
\]

The output is obtained using a **step activation function**:
\[

\text{output} =
\begin{cases}
1, & \text{if } z \ge 0 \\
0, & \text{if } z < 0
\end{cases}
\]

> FORWARD PROPAGATION:

- forward propagation passes input data forward through network layers to produce predictions
- Each layer computes weighted sums \( z = \mathbf{w} \cdot \mathbf{x} + b \) ,  applies activation (e.g., ReLU, sigmoid), and outputs to the next layer.
- The network ends with an output layer prediction \( \hat{y} \), which is compared with the true label \( y \) using a loss function such as Mean Squared Error (MSE).

> BACKWARD PROPAGATION
- Backpropagation computes gradients of loss w.r.t. weights by propagating errors backward.
- Backpropagation uses the **chain rule**. It starts by computing the error at the output layer:

$$
\delta^{L} = \frac{\partial L}{\partial z^{L}}
$$

For any hidden layer \( l \), the error is propagated backward as:

$$
\delta^{l} = (W^{l+1})^{T} \delta^{l+1} \odot \sigma'(z^{l})
$$

 the chain rule is the calculus rule that lets you compute how the loss changes with respect to each weight by decomposing the derivative into a product of simpler partial derivatives across layers.

 - The weights are updated using **gradient descent** 
 which is an **OPTIMIZER**:

$$
\mathbf{w} \leftarrow \mathbf{w} - \eta \, \frac{\partial L}{\partial \mathbf{w}}
$$

\eta → learning rate


The loss function for a single training example is defined as Mean Squared Error (MSE):

$$
L = \frac{1}{2}(y - \hat{y})^2
$$


**Bottom Equation: Chain Rule Decomposition**

The gradient with respect to an old weight is decomposed using the chain rule as:

$$
\frac{\partial L}{\partial w_{\text{old}}}
=
\frac{\partial L}{\partial o_2}
\times
\frac{\partial o_2}{\partial o_1}
\times
\frac{\partial o_1}{\partial w_{\text{old}}}
$$

This decomposes the gradient into successive effects: the influence of the loss \( L \) on the pre-activation \( o_2 \), the propagation from \( o_2 \) to the neuron output \( o_1 \), and the local dependence of \( o_1 \) on the weight \( w_old \).

Forward happens first (compute \( \hat{y} \)); backprop follows (update params).

#### NEURON:
A neuron is a computational usnit that takes one or more inputs, multiplies them by weights, adds a bias then passes the result though an activatiob function

**Mathematical representation of a single neuron:**

$$
z = w_1 x_1 + w_2 x_2 + \cdots + w_n x_n + b
$$

The output of the neuron is:

$$
a = \phi(z)
$$

where \( \phi \) is the activation function.

- Weights are parameters that tell the network “how important” each input is; large positive weight increases the neuron’s output when that input is high, negative weight decreases it.
- Bias is an extra parameter added to the weighted sum that lets the neuron shift its decision boundary, so it doesn’t have to pass through the origin

#### LAYERS:
- <b>Input Layer </b> just holds the features (pixels, tabular features, etc.); it does not do computation
- <b>Hidden Layers </b>sit between input and output and perform most of the computation: each neuron takes outputs from the previous layer, applies weights, bias, and activation.
- <b>Output layers</b> produces final predictions (e.g., class probabilities for classification, a real value for regression)

Deep learning = neural networks with multiple hidden layers.

#### Vanishing gradient problem :

The vanishing gradient problem occurs when gradients become exponentially small during backpropagation in deep neural networks, stalling training in early layers.

core cause:
- Chain rule multiplies many small derivatives (e.g., sigmoid/tanh gradients ≤ 0.25) across layers
- Product shrinks rapidly, making early-layer weights barely update

effects:
- Early layers learn slowly or stop, capturing only basic features while later layers overfit.
- Deep plain networks (pre-2012) failed; perceptron-like shallow nets worked better.
