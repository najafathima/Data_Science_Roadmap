### OPTIMIZERS:

An optimizer is an algorithm used to minimize the loss function by updating the model parameters (weights and biases) during training.

Without optimizers, the model cannot learn.

their job is to :
- Minimize loss
- Speed up convergence
- Avoid local minima
- Stabilize training
- Handle noisy gradients


## Types of Optimizers in Deep Learning

Optimizers are algorithms used to update neural network weights in order to minimize the loss function during training.

### 1️⃣ First Generation Optimizers

These are the basic gradient-based optimization methods.

- **Gradient Descent**

### 2️⃣ Improved Gradient Descent

Methods that improve convergence speed and stability.

- **Stochastic Gradient Descent (SGD)**
- **Mini-batch Gradient Descent**

### 3️⃣ Momentum-Based Optimizers

These optimizers accelerate training by incorporating past gradients.

- **Momentum**
- **Nesterov Accelerated Gradient (NAG)**

### 4️⃣ Adaptive Learning Rate Optimizers

These methods automatically adjust the learning rate for each parameter.

- **AdaGrad**
- **RMSProp**
- **Adam**
- **AdamW**

### EPOCH:

An epoch is one complete pass of the entire training dataset through the neural network

eg: dataset size = 1000 samples

- epoch 1 : model sees all 1000 sample once
- epoch 10 : dataset has been processed 10 times

neural network cannot learn patterns in one pass. Repeated passes allow:
- Gradual weight adjustment
- Reduction of loss
- Better pattern learning

### ITERATION:

An iteration is one update of model parameters.

### BATCH SIZE:

Batch size is the number of training samples used to compute one gradient update



>#### Gradient Descent :

Gradient Descent is an optimization algorithm that minimizes a loss function by moving the weights in the direction opposite to the gradient.

**Gradient Descent Weight Update**

$$
w_{\text{new}} = w_{\text{old}} - \eta \frac{\partial L}{\partial w_{\text{old}}}
$$

Intuition:

Imagine you are standing on a mountain and want to reach the lowest valley.

The gradient tells you:

"Which direction is downhill?"

You move in that direction step by step.

problem:

Gradient descent can be:
- slow
- stuck in local minima
- computationally expensive

because it **uses entire dataset** to compute gradients


Since the gradient is computed using all samples, the update happens once per epoch.

>#### STOCHASTIC GRADIENT DESCENT:

Stochastic Gradient Descent (SGD) updates the model parameters using one training example at a time.

Weight Update Rule (Stochastic Gradient Descent)

In Stochastic Gradient Descent (SGD), the weights are updated using the gradient computed from a **single training example**.

$$
w_{\text{new}} = w_{\text{old}} - \eta \frac{\partial L^{(i)}}{\partial w_{\text{old}}}
$$

\partial L^{(i)} - Loss of the i-th trainig example

SGD perform many updates per EPOCH

Advantages:
- faster learning
- can escape local minima

Disadvantage:
- very noisy updates
- oscillating convergence

> ### MINI BATCH GRADIENT DESCENT:

Mini-Batch gradient descent updates weights using a small subset of the dataset

weight Update Rule (Mini-batch Gradient Descent)

In Mini-batch Gradient Descent, the gradient is computed using a small batch of training examples instead of the entire dataset.

$$
\mathbf{w} = \mathbf{w} - \eta \nabla L_{\text{batch}}(\mathbf{w})
$$

Mini-batch GD is the most commonly used method in deep learning because it balances speed and stability.
- Faster than batch GD
- More stable than SGD
- Efficient GPU computation

> ### SGD with MOMENTUM

SGD with momentum is an improved versionof SGD that adds a memory of previous gradients to accelerate learning and reduce oscillations during optimization

In normal SGD, the optimizer can behave like:
- Updtes are very noisy
- It zig zags when the loss surface is steep
- Convergence become slow

**momentum helps the optimizer move faster toward the global minimum.**

Momentum accelerates gradient descent by accumulating past gradients.

#### Step 1 — Velocity Update

$$
v_t = \gamma v_{t-1} + \eta \nabla L(w_t)
$$

where

\( v_t \) = velocity at step \( t \)  
\( \gamma \) = momentum coefficient (typically 0.9)  
\( \eta \) = learning rate  
\( \nabla L(w_t) \) = gradient of the loss with respect to weights


#### Step 2 — Weight Update

$$
w_{t+1} = w_t - v_t
$$

Advantages:
- Reduce noise
- Works with mini batch
- Quicker convergence

> ### ADADELTA :

AdaDelta is an adaptive learning rate optimizer thatimproves AdaGrad by preventing the learning rate (by controlling αt) from becoming extremely small during training

in adagrad , the accumulated squared gradients is stored in :
$$
\alpha_t = \sum_{i=1}^{t} g_i^2
$$

in AdaDelta instead of αt​=∑gi2​ we use :

$$
s_{dw,t} = \rho s_{dw,t-1} + (1-\rho) g_t^2
$$

- ρ = decay factor (usually 0.9)

sdw​ is the exponential moving average of squared gradients.

It stores recent gradient magnitudes instead of summing all past gradients like AdaGrad.

so , Weight update in AdaDelta:

$$
w_{t+1} =
w_t -

{\sqrt{s_{dw,t}+\epsilon}} g_t

> ### RMSProp:

RMSProp (Root Mean Square Propagation) is an optimizer that adapts the learning rate by dividing the gradient by the root mean square of recent squared gradients.

Instead of accumulating all past gradients (like AdaGrad), RMSProp uses an exponentially decaying moving average.

Moving Average of Squared Gradients

$$
S_t = \beta S_{t-1} + (1 - \beta) g_t^2
$$

where

\( S_t \) = running average of squared gradients  
\( \beta \) = decay rate (typically around \(0.9\))  
\( g_t \) = gradient at step \(t\), where

$$
g_t = \nabla L(w_t)
$$

 Weight Update

$$
w_{t+1} =
w_t -
\frac{\eta}{\sqrt{S_t + \epsilon}} \, g_t
$$


advantages:
- prevents learning rate from becoming too small
- faster convergence
- good for deep neural network
- handles non-stationary problems



>### Adam Optimizer:

Adam (Adaptive Moment Estimation) is an optimization algorithm used to train neural networks that combines the ideas of Momentum and RMSProp to adaptively adjust the learning rate for each parameter during training.

It keeps track of:

1. First moment (mean of gradients) → like momentum

2. Second moment (variance of gradients) → like RMSProp

This allows Adam to achieve fast convergence and stable training, which is why it is one of the most widely used optimizers in deep learning.
