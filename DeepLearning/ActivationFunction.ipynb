## Activation Function:

Activation function are where neural network become powrful.

without activation function -> neural networks = just linear regression.

with activation -> it can learn complex nonlinear patterns.

An activation function \( \phi(\cdot) \) is a non-linear function applied to the neuron's weighted sum \( z \) to produce the output \( a \).

without activation functions (using only linear functions), stacking layers collapses to a single linear transformation, so the network cannot learn non‑linear patterns.

> ### SIGMOID ACTIVATION FUNCTION

The Sigmoid function is a smooth, S-shaped activation function that maps any real-valued number into the range:  (0,1)

It is mainly used in binary classification output layers.

**Mathematical Equation (Sigmoid Activation Function)**

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

where

$$
z = \mathbf{w}^{T}\mathbf{x} + b
$$

and \( e \) is Euler’s constant (approximately 2.718).

gives result as 0 or 1 

if y >= 0.5 =1

else =0

<img src="sigmoid.png" width=400/>

the derivative curve will always be ranging between 0-0.25

use when:
- Binary classification output layer
- when the output must be probability

problem:
- suffers from vanishing gradient problem
- not zero centered

> ### TanH (HYPERBOLIC TANGENT) ACTIVATION FUNCTION:
Tanh is a scaled and shifted version of sigmoid that outputs values between -1 and 1.

it is zero centered , smooth and differentiable anywhere

Equation:

$$
\tanh(z) = \frac{e^{z} - e^{-z}}{e^{z} + e^{-z}}
$$

<img src='tanh.png' width=400/>

usage :
-  Older RNNs
- Hidden layers 

problem:
- Still suffers from vanishing gradient problem

> ### RELU(Rectified Linear Unit):
ReLU outputs the input directly if positive; otherwise outputs zero.

Equation:
$$
f(z) = \max(0, z)
$$

derivative:

$$
f'(z) =
\begin{cases}
1, & \text{if } z > 0 \\
0, & \text{if } z \le 0
\end{cases}
$$

<img src='ReLU.png' width=400/>

use:
- most widely used activation for hidden layers

problem:
- Dying ReLU problem
-> if neuron outputs negative always -> gradient=0 -> neuron never updates

advantages:
- Solves vanishing gradient problem (for positive region)
- Computationally cheap
- Speeds up training

> ### LEAKY ReLU Function

Leaky ReLU allows a small slope for negative value intead of zero

$$
f(z) =
\begin{cases}
z, & \text{if } z > 0 \\
\alpha z, & \text{if } z \le 0
\end{cases}
$$
Here, \( \alpha \) is a small positive constant (typically 0.01).

output range:
(-∞,∞)

<img src='Leaky-ReLU.webp' width=400/>

advantage:
- Solve dead neuron problem in ReLU

> ### ELU (Exponential Linear Unit):
ELU is an activation function that behaves like ReLU for positive values, but instead of becoming zero for negative values, it smoothly decays exponentially toward a negative saturation value.
$$
f(z) =
\begin{cases}
z, & \text{if } z > 0 \\
\alpha \left(e^{z} - 1\right), & \text{if } z \le 0
\end{cases}
$$

Here, \( \alpha \) is a positive constant that controls the value for negative inputs.

<img src='elu.png' width=400/>

introduced to:
- fix dying relu
- reduce bias shift
- improve learning speed

> ### PReLU 

PReLU is an extension of Leaky ReLU where the slope of the negative region is not fixed, but learned during training.

<img src='prelu.png' width=400/>

Mathematical equaion:

$$
f(z) =
\begin{cases}
z, & \text{if } z > 0 \\
\alpha z, & \text{if } z \le 0
\end{cases}
$$

α is learned during training

nature:

- non -linear
- piecewise linear
- not smootha t 0
- allows negative outputs
- adaptive behavior

advantages:
- avoids dying ReLU
- learns optimal slope
- better performance in deep cnn

TECHNIQUE TO UNDERSTAND WHICH ACTIVATION FUNCTION WE SHOULD USE:

- BINARY CLASSIFICATION -> ReLU(hidden layer) , Sigmoid(in Output layer)
- MULTIPLE CLASSIFICATION -> ReLU(in Hidden layer) , Softmax(in Output layer)
- REGRESSION -> ReLU(in Hidden Layer), Linear Activation(in Output Layer)


### LOSS FUNCTIONS:

A loss function measures how wrong a single prediction is compared to the true value.

### COST FUNCTION :
A cost function is the average (or total) of loss over the entire dataset.

**Loss vs Cost**

| Loss | Cost |
|------|------|
| Computed for one training example | Computed over the entire dataset |
| Used in theoretical explanations | Used during optimization |
| Represents local error | Represents global error |

For a single example:

$$
L = \frac{1}{2}(y - \hat{y})^2
$$

For the entire dataset of \( m \) examples:

$$
J = \frac{1}{m} \sum_{i=1}^{m} L^{(i)}
$$



##### REGRESSION:
- MSE (Mea Squared Error)
- MAE (Mean Absolute Error)
- HUBER LOSS

> MEAN SQUARED ERROR (MSE):

Measures average squared difference between prediction and true value.


#### Loss Function (for one training example)

For a single example:

$$
L = \frac{1}{2}(y - \hat{y})^2
$$

where  
\( y \) = true label  
\( \hat{y} \) = predicted output  

The factor \( \frac{1}{2} \) is added for mathematical convenience (it cancels out during differentiation).

---

#### Cost Function (for entire dataset)

For a dataset with \( m \) training examples:

$$
J = \frac{1}{2m} \sum_{i=1}^{m} \left( y^{(i)} - \hat{y}^{(i)} \right)^2
$$

where  
\( m \) = number of training examples.

- Penalizes large error heavily
- Differentiable
- Sensitive to outliers - disadvantage

> MEAN ABSOLUTE ERROR (MAE):

Measures absolute difference

#### Loss Function (for one training example)

For a single example:

$$
L = |y - \hat{y}|
$$

where  
\( y \) = true label  
\( \hat{y} \) = predicted output  

---

#### Cost Function (for entire dataset)

For a dataset with \( m \) training examples:

$$
J = \frac{1}{m} \sum_{i=1}^{m} \left| y^{(i)} - \hat{y}^{(i)} \right|
$$

where  
\( m \) = number of training examples.

- Robust to outliers - overcame the issue with mse
- not differentiable at 0
- Gradient constant -> slow convergence. - disadvantage

> HUBER LOSS:

Huber Loss is a hybrid loss function that behaves like MSE for small errors and like MAE for large errors

it combines:
- the smoothness of MSE
- the robustness of MAE

🔹 **Mathematical Definition (Huber Loss)**

Let

$$
e = y - \hat{y}
$$

Huber Loss is defined as:

$$
L(e) =
\begin{cases}
\frac{1}{2} e^2, & \text{if } |e| \le \delta \\[6pt]
\delta \left( |e| - \frac{1}{2}\delta \right), & \text{if } |e| > \delta
\end{cases}
$$

where δ is a threshold hyperparameter that controls where the function switches behavior.

- When error is **small** -> behaves like **MSE**
- When error is **large** -> behaves like **MAE** 

ADVANTAGES:

- Smooth near zero -> good for gradient descent
- Less sensitive to outliers than MSE
- Differentiable everywhere

Loss and Cost functions are differentiable because:

neural network are trained using gradient-based optimization, and gradients required derivatives

if loss is not differentiable:
- gradient becomes undefined
- backpropogation breaks
- optimization become unstable

### CLASSIFICATION:

> BINARY CROSS ENTROPY:

Binary Cross Entropy measures the difference between true binary labels(0 or 1) and predicted probabilities

Used for binary classifiction problems

**Binary Cross-Entropy (BCE) Loss**

$$
L = - \left[ y \log(p) + (1 - y)\log(1 - p) \right]
$$

where  

y = true label (0 or 1)  
p = predicted probability of the positive class.

properties:

- Works well with **SIGMOID** Activation
- Strong gradient when prediction is wrong
- widely used in Logistic regression and neural network


> CATEGORICAL CROSS ENTROPY:

Used when there are multiple classes (>2).

**Categorical Cross-Entropy Loss**

$$
L = - \sum_{i=1}^{C} y_i \log(p_i)

$$

where

\( C \) = number of classes  
\( y_i \) = true label (one-hot encoded)  
\( p_i \) = predicted probability for class \( i \)

properties:

- Works with **SOFTMAX** Activation
- Used for multiclass classification
- penalizes confident wrong prediction heavily
