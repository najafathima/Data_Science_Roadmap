### <u><B> *NATURAL LANGUAGE PROCESSING (NLP)*</B></U>

NLP is a subfield of liguistics , computer science , and Artificial intelligence concerned with the interactions between computers and human language, in particular how to program computers to process and analyse large amounts of natural language data

OR 

Teaching machines how humans talk, write, and mean things.

##### APPLICATIONS:

- Contextual Advertisements
- Email Clients - spam filtering, smart reply
- Social media - removing adult content, opinion mining
- Search engines
- Chatbots and virtual assistance


##### Common NLP tasks:

- Text classification
- Tokenization
- Text preprocessing
- Named entity recognition
- Sentiment analysis
- Machine Translation
- Text generation
- Question answering


##### Approaches to NLP

<u>Heuristic Methods</u>
- Using manually designed rules
- Based on linguistic intuition
- Without training on large datasets
<u>Machine Learning Based Methods</u>
- 👉 Instead of saying “if word = good → positive”,
- 👉 we say “learn from thousands of labeled examples what ‘positive’ looks like.”
- 
<u>Deep Learning Based Methods</u>

Architectures used:
- RNN
- LSTM
- GRU/CNN
- Transformers


##### Challenges in NLP:
- Ambiguity : may be machine can't recognise the exact meaning what a sentence carries
- Context dependence : sarcastic sentences can be misunderstood as genuine
- Multiple languages and dilects
- Noisy data
- Bias in data

TOKENIZATION:

Tokenization is the process of breaking text into smaller units called tokens so that a computer can process language.

"I love NLP!" -> ["I", "love", "NLP", "!"]

<u> Tokenization algorithms:</u>

Byte pair encoding(BPE)

WordPiece 

SentencePiece
