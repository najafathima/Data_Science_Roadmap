### <u><B> *REINFORCEMENT LEARNING*</B></U>

Reinforcement Learning = learning by interaction + feedback

Instead of learning from labeled data(supervised) or structured (unsupervised), an agent learns by doing and getting rewards or penalties

eg:
- A child learning to ride a cycle
- A game bot learning to win
- A robot learning to walk

Core components of Reinforcement Learning

> AGENT
- The leader / decision maker
- eg: robot,game player, trading bot

> ENVIRONMENT
- The world the agent interacts with
- eg: game board, road, market

> STATE (S)
- Current situation of the agent
- eg: position of robot , game score

> ACTION (A)
- What the agent can do
- eg: move left/right, buy/sell

> REWARD (R)
- Feedback from the environment 
- eg: +10 for winning , -1 for losing

The learning loop is: 

**observe state → choose action → environment changes → get reward → update policy/values → repeat**
 ,balancing exploration vs exploitation.

##### MARKOV DECISION PROCESS:
A Markov Decision Process is a mathematical way to describe any reinforcement learning problem.

If a problem can be written as an MDP, we can apply RL to it.

An MDP is defined as:

⟨𝑆,𝐴,𝑃,𝑅,𝛾⟩

1. STATE (S): All possible situaions the agent can be in at a given time
2. ACTIONS (A) : All possible actions the agent can take
3. TRANSITION PROBABILITY (P) : Probability of going to next state s′
given current state s and action a P(s′∣s,a)
4. REWARD (R) : Immediate feedback after an action. R(s,a,s′)
5. DISCOUNT FACTOR (γ) : Controls how much the agent cares about the future. 0≤γ≤1 
- γ = 0 → only present reward matters

- γ ≈ 1 → future rewards matter a lot

##### VALUE FUNCTION (V(s)):
It measures the expected total reward an agent can collect from now onward, when a perticular state is chosen

limitation : V(s) only tells how good a state is - but not which action to take

##### ACTION VALUE FUNTION (Q-function  Q(s,a)):
Expected future reward when
you take action a in state s,
and then behave optimally.

V(s) only evaluates states, but it does not distinguish between different actions within the same state. To choose the best action, we need the action-value function Q(s, a).

##### Value-based methods (Q-learning)

Value-based methods learn “how good” each action is in each state. The most famous is Q-learning, which learns Q(s,a) (expected return after taking action a in state s)
