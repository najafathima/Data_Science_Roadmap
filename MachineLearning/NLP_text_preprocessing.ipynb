#### 2. REMOVE HTML TAGS

```python
import re
def remove_html_tags(text):
    pattern=re.compile("<.*?>")
    return pattern.sub(r'',text)
```

In Python, the re library stands for Regular Expressions.

It is used for pattern matching and text processing.

without .*? is greedy , It matches as much as possible

r'' → replace with empty string

```python
text="<html><body><h1>Breaking News!</h1><p>This <b>movie</b> was <i>absolutely amazing</i> 😍.</p><br><p>Visit <a href='https://example.com'>our site</a> for more <span style='color:red;'>updates</span>.</p><!-- end of article --></body></html>"
remove_html_tags(text)
```

#### 3.REMOVE URLs

```python
def remove_url(text):
    pattern= re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

http -> matches literal 

s? -> means optional , matches both http and https

:// -> matches literal

\S+ -> one or more non white space characters

| -> or , the second part 

```python
text="Check this article https://example.com and visit www.test.com for more info."
remove_url(text)
```

#### 4. REMOVE PUNCTUATION


```python
import string,time
string.punctuation
```

```python
exclude=string.punctuation
```

```python
def remove_punc(text):
    for char in exclude:
        text=text.replace(char,'')
    return text
```

```python
text= "{String with punctuation??/;}"
start = time.time()
print(remove_punc(text))
time1=time.time() - start #to measure how much tym will take to complete the entire process
print(time1)
```

this will take a lot of time

so an another technique can be used

```python
def remove_punc1(text):
    return text.translate(str.maketrans('','',exclude))
```

str.maketrans(x, y, z)

Where:

- x → characters to replace

- y → characters to replace with

- z → characters to delete

str.maketrans('','',exclude)

Meaning:

- Replace nothing

- With nothing

- Delete everything inside exclude

translate() applies that translation table to the string.

```python
start= time.time()
remove_punc1(text)
time2=time.time()-start
print(time2)
```

```python
time1/time2

# We can find that the second function is 6 times faster than the first one
```

```python
df['review']=df['review'].apply(remove_punc1)
df
```

#### 7. REMOVING STOP WORD 

Words like a, the, of, are, my etc are contributed to sentence formation but not in the meaning of the sentence

Removing common, less informative words from text before processing.

So, it won't be useful in sentiment analysis or any kind of things like that, so we need to remove it

In machine learning we need this to:

- Reduce vocabulary size
- Reduce noise
- Improve model efficiency
- Reduce dimentionality
- Improve TF-IDF performance


```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text=[]

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)

    x = new_text[:]
    new_text.clear()
    return ' '.join(x)
```

```python
remove_stopwords("NLTK tries to load the stopwords corpus file from your system.")
```

#### 7. HANDLING EMOJIS



<U>*METHOD 1: USING <b>REGEX</b>*</U>

```python
import re

def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map
        u"\U0001F1E0-\U0001F1FF"  # flags
        u"\U00002700-\U000027BF"
        u"\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'',text)
```

```python
text= "I love NLP 😍🔥🚀"
print(remove_emoji(text))
```

### <u><b>STEMMING</b> </u>

The process of reducing a word to its root (base) form by cutting off suffixes.

it removes endings like:
- -ing
- -ed
- -ly
- un- 
- -s etc

```python
from nltk.stem.porter import PorterStemmer

ps=PorterStemmer()
def stem_words(text):
    return ' '.join([ps.stem(word) for word in text.split()])
```

```python
sample= ' walk walks walking walked'
stem_words(sample)
```

```python
text= """
Dr. John Smith lives in the U.S.A. He works at Google Inc. as a senior AI researcher.
Every morning, he wakes up at 6:30 a.m., drinks coffee, and reads the news.
"Artificial Intelligence is the future!" he often says.
Last week, he visited New York. It was amazing, wasn't it?
He met Prof. Williams there, and they discussed NLP, ML, and robotics.
Can machines really think like humans? That is the big question.
"""

stem_words(text)
```

### <u><b>LEMMATIZATION</b> </u>

Lemmetization, unike Stemming , reduces the inflected words properly ensuring that the root word belongs to the language . In lemmetization root word is called Lemma. A lemma is the canonical form, dictionary form, criterion form os a set of words

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer= WordNetLemmatizer()

sentence= 'I love NLP. It is amazing! spacy is powerful, isn,t it?'
punctuations= "?:!.,;"
sentence_words=nltk.word_tokenize(sentence,preserve_line=True) #This splits the sentence into tokens (words and punctuation).
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)
#sentence_words
print('{0:20}{1:20}'.format('Word','Lemma'))
for word in sentence_words:
    print('{0:20}{1:20}'.format(word,wordnet_lemmatizer.lemmatize(word)))
```
