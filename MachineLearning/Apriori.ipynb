### <u><B> *APRIORI ALGORITHM*</B></U>

Apriori is an unsupervised learning algorithm used for 

-> ASSOCIATION RULE MINING

Apriori finds patterns like :

"People who buys X often also buy Y

It works on TRANSACTIONAL DATA 

```python
import pandas as pd


transactions = [
    ['Beer','Diaper'],
    ['Milk', 'Diaper', 'Bread','Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter'],
    ['Beer','Cookies', 'Diaper']
]

transactions
```

Convert transactions → one-hot encoded DataFrame

```python
from mlxtend.preprocessing import TransactionEncoder

te=TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df=pd.DataFrame(te_array,columns=te.columns_)
df
```

Apply apriori algorithm

```python
from mlxtend.frequent_patterns import apriori

freq_itemsets=apriori(
    df,
    min_support=0.4,
    use_colnames=True
)

freq_itemsets
```

Generate association rule:


```python
from mlxtend.frequent_patterns import association_rules

rules=association_rules(
    freq_itemsets,
    metric='confidence',
    min_threshold=0.6
)

rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```
