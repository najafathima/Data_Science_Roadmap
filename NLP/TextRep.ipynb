##### Questions to get answered first

1. What is feature extraction from text?

Feature extraction from text is the process of converting raw textual data into numeircal representation(features) so that machine learning models can preocess and learn from it

Feature extraction = Converting text → meaningful numerical vectors.

2. Why do we need it?

- Machines don't understand the text
- To identify patterns
- To reduce ambiguity


3. Why is it difficult?

Text data is complex

- Unstructured Nature
- High dimensionality
- Context and polysemy - Same word different meaning
- Word order matters - I love NLP vs NLP loves me

4. What is the core idea?

The core idea of text feature extraction is :

Representing text in a mathematical vector space such that similar texts are close together and different texts are far apart

5. What are the techniques?

- BAG of WORDS (BoW)
- N-grams
- TF-IDF
- Word embeddings
- Contextual embeddings

#### BAG of Words:
Bag of Words is a text representation technique that converts text into numerical vectors by counting the occurance of each word in a document

It ignores:
- grammer
- word order
- sentence structure

It cares about :
- whether a word appears 
- how many times it appears

"I love NLP"
"NLP love I"

both gives same representation


```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus= [
    'I love NLP',
    'NLP is amazing',
    'I love Machine learning'
]

vectorizer= CountVectorizer()

bow=vectorizer.fit_transform(corpus)

print('vocabulary: ', vectorizer.get_feature_names_out())
print('Bow matrix:\n ',bow.toarray())
print(vectorizer.vocabulary_)
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
```

How BoW works internally:
- Lowercasing
- Tokenization
- Remove punctuation
- Build vocabulary
- Count frequencies
- Create sparse matrix

LIMITATIONS:
- Ignores word order
- ignores context
- high dimensionality
- No semantic meaning

#### TF-IDF
TF-IDF (Term Frequency – Inverse Document Frequency) is a statistical measure used to evaluate how important a word is to a document relative to a corpus.

It assigns a weight to each word in each document.

the weight increases when:
- The word appears frequently in a document
- The word appears rarely across the corpus

the weight decreases when:
- The word appears in many documents in the corpus

#### - Term Frequency(TF) measures :
How frequently a word appears in a document

$$
TF(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}
{\text{Total number of terms in document } d}
$$

#### - Inverse Document Frequency(IDF) measures :

How rare a word is across all documents in the corpus

If word appears in:

- Every document → IDF small

- Few documents → IDF large


```python
import pandas as pd

data = {
    "Doc_ID": [1, 2, 3, 4],
    "Text": [
        "I love this movie",
        "This movie is amazing",
        "I hate this movie",
        "This film is boring"
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer= TfidfVectorizer()
X= vectorizer.fit_transform(df['Text'])

print('vocabulary: ', vectorizer.vocabulary_)
```

```python
print('TF_IDF matrix: \n', X.toarray())
```

TfidfVectorizer() internally:

- Converts text to lowercase

- Tokenizes

- Removes punctuation

- Builds vocabulary

- Calculates TF

- Calculates IDF

- Multiplies TF × IDF

- Returns sparse matrix
